Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
right_coord = [28.56367,-80.57163]
left_coord = [34.63607,-120.62503]
# e.g.,: Lat: 28.56367  Lon: -80.57163
unique_df = spacex_df.drop_duplicates(subset="Launch Site", keep="first")
unique_df
df_right = unique_df[ unique_df["Launch Site"].isin(['CCAFS LC-40','KSC LC-39A','CCAFS SLC-40']) ]
df_left = unique_df[ unique_df["Launch Site"].isin(['VAFB SLC-4E']) ]
df_right.head()


distance_right = []
distance_left = []

for i,record in df_right.iterrows():
    
    distance_right.append(calculate_distance(record['Lat'],record['Long'],right_coord[0], right_coord[1]))
    
for i,record in df_left.iterrows():
    
    distance_left.append(calculate_distance(record['Lat'],record['Long'],left_coord[0], left_coord[1]))
print(distance_right)
print(distance_left)
```

```python
# Add distance labels at the center of each launch-site-to-coastline line

# Right-side launch sites
for _, record in df_right.iterrows():
    start = [record['Lat'], record['Long']]
    end = right_coord
    mid = [(start[0] + end[0]) / 2, (start[1] + end[1]) / 2]
    dist = calculate_distance(start[0], start[1], end[0], end[1])

    folium.Marker(
        location=mid,
        icon=DivIcon(
            icon_size=(120, 20),
            icon_anchor=(60, 10),
            html=f'<div style="font-size: 12px; color:#d35400;"><b>{dist:.2f} KM</b></div>'
        )
    ).add_to(site_map)

# Left-side launch sites
for _, record in df_left.iterrows():
    start = [record['Lat'], record['Long']]
    end = left_coord
    mid = [(start[0] + end[0]) / 2, (start[1] + end[1]) / 2]
    dist = calculate_distance(start[0], start[1], end[0], end[1])

    folium.Marker(
        location=mid,
        icon=DivIcon(
            icon_size=(120, 20),
            icon_anchor=(60, 10),
            html=f'<div style="font-size: 12px; color:#d35400;"><b>{dist:.2f} KM</b></div>'
        )
    ).add_to(site_map)

site_map
```

*TODO:* Draw a `PolyLine` between a launch site to the selected coastline point


```python
# Draw PolyLines for right-side launch sites
for i, record in df_right.iterrows():

    folium.PolyLine(
        locations=[
            [record['Lat'], record['Long']],     # launch site
            right_coord                           # closest coastline point
        ],
        weight=2,
        color='blue'
    ).add_to(site_map)


# Draw PolyLines for left-side launch site(s)
for i, record in df_left.iterrows():

    folium.PolyLine(
        locations=[
            [record['Lat'], record['Long']],     # launch site
            left_coord                            # closest coastline point
        ],
        weight=2,
        color='green'
    ).add_to(site_map)


site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Example launch site (CCAFS LC-40)
launch_site = launch_sites_df[launch_sites_df['Launch Site'] == 'CCAFS LC-40'].iloc[0]
launch_coord = [launch_site['Lat'], launch_site['Long']]

# Example nearby points (update with your MousePosition coordinates)
proximities = {
    'Closest City': [28.4019, -80.6043],
    'Closest Railway': [28.5721, -80.5853],
    'Closest Highway': [28.5633, -80.5708],
}

for name, point_coord in proximities.items():
    # Create a marker with distance to a closest city, railway, highway, etc.
    distance_km = calculate_distance(
        launch_coord[0], launch_coord[1],
        point_coord[0], point_coord[1]
    )

    folium.Marker(
        location=point_coord,
        popup=f'{name}: {distance_km:.2f} km',
        icon=folium.Icon(color='purple', icon='info-sign')
    ).add_to(site_map)

    # Draw a line between the marker to the launch site
    folium.PolyLine(
        locations=[launch_coord, point_coord],
        weight=2,
        color='purple'
    ).add_to(site_map)

    # Optional: distance label at midpoint
    midpoint = [(launch_coord[0] + point_coord[0]) / 2, (launch_coord[1] + point_coord[1]) / 2]
    folium.Marker(
        location=midpoint,
        icon=DivIcon(
            icon_size=(120, 20),
            icon_anchor=(60, 10),
            html=f'<div style="font-size: 12px; color:purple;"><b>{distance_km:.2f} KM</b></div>'
        )
    ).add_to(site_map)

site_map
```
